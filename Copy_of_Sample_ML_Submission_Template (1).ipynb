## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Merge datasets on common column 'id'

try:
    df = titles.merge(credits, on='id', how='left')
    print("Datasets merged successfully ✅")
    print("Merged shape:", df.shape)
except Exception as e:
    print("Error during merging:", e)

# Remove duplicate rows

df = df.drop_duplicates()

# Reset index
df.reset_index(drop=True, inplace=True)

print("After removing duplicates:", df.shape)

# Check missing values

df.isnull().sum()

# Handle important missing values

df['description'] = df['description'].fillna('')
df['genres'] = df['genres'].fillna('[]')
df['production_countries'] = df['production_countries'].fillna('[]')

import ast

# Safe conversion function
def convert(text):
    try:
        return ast.literal_eval(text)
    except:
        return []

# Convert to list
df['genres_list'] = df['genres'].apply(convert)
df['countries_list'] = df['production_countries'].apply(convert)

# Create new features
df['genre_count'] = df['genres_list'].apply(len)
df['country_count'] = df['countries_list'].apply(len)
df['desc_length'] = df['description'].apply(lambda x: len(str(x).split()))

df[['genre_count', 'country_count', 'desc_length']].head()

df.head()
````

````python
# Keep only one row per movie (remove duplicates based on id)

df = df.drop_duplicates(subset='id')

# Reset index
df.reset_index(drop=True, inplace=True)

print("After fixing duplicate movie entries:", df.shape)
````

The datasets were merged using the common column 'id' to create a unified dataset.

Feature engineering was performed by converting string representations into lists and creating new meaningful features such as genre count, country count, and description length.

These transformations help in simplifying the dataset and improving the performance of machine learning models.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Based on exploratory data analysis, three hypotheses are formulated to understand the relationship between different features and IMDB score.

These hypotheses focus on content type, runtime, and genre influence on ratings. Statistical tests will be used to validate whether the observed differences are significant or due to random variation.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H0 (Null Hypothesis): There is no significant difference in IMDB scores between movies and TV shows.

H1 (Alternate Hypothesis): There is a significant difference in IMDB scores between movies and TV shows.

#### 2. Perform an appropriate statistical test.

````python
# Hypothesis 1: Movies vs TV Shows (IMDB Score Comparison)

from scipy.stats import ttest_ind

# Separate groups
movies = df[df['type'] == 'MOVIE']['imdb_score'].dropna()
shows = df[df['type'] == 'SHOW']['imdb_score'].dropna()

# Perform t-test (Welch’s t-test - does not assume equal variance)
t_stat, p_value = ttest_ind(movies, shows, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis → Significant difference exists")
else:
    print("Fail to Reject Null Hypothesis → No significant difference")
````

##### Which statistical test have you done to obtain P-Value?

An Independent Two-Sample t-test (Welch’s t-test) has been performed to compare the mean IMDB scores between movies and TV shows.Answer Here.

##### Why did you choose the specific statistical test?

The Independent t-test is appropriate because we are comparing the means of a numerical variable (IMDB score) between two independent groups (movies and TV shows).

Welch’s t-test is used specifically because it does not assume equal variance between the two groups, making it more suitable for real-world datasets.Answer Here.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H0 (Null Hypothesis): There is no significant relationship between runtime and IMDB score.

H1 (Alternate Hypothesis): There is a significant relationship between runtime and IMDB score.

#### 2. Perform an appropriate statistical test.

````python
# Hypothesis 2: Runtime vs IMDB Score

from scipy.stats import pearsonr

# Drop missing values
runtime = df['runtime'].dropna()
ratings = df['imdb_score'].dropna()

# Align indices (IMPORTANT)
common_index = runtime.index.intersection(ratings.index)

runtime = runtime.loc[common_index]
ratings = ratings.loc[common_index]

# Perform Pearson correlation test
corr, p_value = pearsonr(runtime, ratings)

print("Correlation:", corr)
print("P-value:", p_value)

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis → Significant relationship exists")
else:
    print("Fail to Reject Null Hypothesis → No significant relationship")
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test has been used to measure the relationship between runtime and IMDB score.

##### Why did you choose the specific statistical test?

The Pearson correlation test is appropriate because both runtime and IMDB score are continuous numerical variables.

It helps measure the strength and direction of the linear relationship between these two variables.Answer Here.

H1 (Alternate Hypothesis): There is a significant relationship between genre count and IMDB score.

#### 2. Perform an appropriate statistical test.

````python
# Hypothesis 3: Genre Count vs IMDB Score

from scipy.stats import pearsonr

# Drop missing values
genre = df['genre_count'].dropna()
ratings = df['imdb_score'].dropna()

# Align indices
common_index = genre.index.intersection(ratings.index)

genre = genre.loc[common_index]
ratings = ratings.loc[common_index]

# Perform Pearson correlation test
corr, p_value = pearsonr(genre, ratings)

print("Correlation:", corr)
print("P-value:", p_value)

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis → Significant relationship exists")
else:
    print("Fail to Reject Null Hypothesis → No significant relationship")
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test has been used to measure the relationship between genre count and IMDB score.Answer Here.

##### Why did you choose the specific statistical test?

The Pearson correlation test is appropriate because both genre count and IMDB score are numerical variables.

It helps determine whether a linear relationship exists between the number of genres and movie ratings.

### 2. Handling Outliers

````python
# Handling Outliers using IQR method

def remove_outliers(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return df[(df[col] >= lower) & (df[col] <= upper)]

# Apply on important columns
df = remove_outliers('runtime')
df = remove_outliers('imdb_score')

print("Outliers handled successfully ✅")
print("New shape:", df.shape)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Outliers were handled using the Interquartile Range (IQR) method.

Values lying outside 1.5 times the IQR were removed to reduce the impact of extreme values on the model.

This helps improve model performance by ensuring that predictions are not biased due to abnormal data points.
